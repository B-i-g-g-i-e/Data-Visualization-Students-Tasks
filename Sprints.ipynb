Data Cleaning: Because the dataset contained zero missing values, we focused data cleaning to ensuring gpa had only a maximum of 4.0

```python
#Checking for anomalies

weird_gpas = df[(df['GPA'] > 4.0) | (df['GPA'] < 0.0)]

# Print out how many weird GPAs we found
print(f"No. of weird GPAs found: {len(weird_gpas)}")

```

there were no weird anomalies in the data

```python
# in case of a number >4.0 or <0 clip() will force the number to 4 or 0 depending on its range
df['GPA'] = df['GPA'].clip(upper=4.0, lower=0.0)

```

```python
# check if it worked
print(f"New Max GPA: {df['GPA'].max()}")
print(f"New Min GPA: {df['GPA'].min()}")
```

```python
# Feature engineering
# creating new data from old data

df['Total_Activities'] = df['Extracurricular'] + df['Sports'] + df['Music'] + df['Volunteering']

print(df['Total_Activities'])
```

```python
df['under_performing'] = np.where((df['Absences'] > 15) | (df['GPA'] < 2.0), 1, 0)
print(df['under_performing'])
```

```python
#removing the floats and creating three portions to work with
df['Study_Habits'] = pd.qcut(df['StudyTimeWeekly'], q=3, labels=['Low', 'Medium', 'High'])

print(df['Study_Habits'])
```
