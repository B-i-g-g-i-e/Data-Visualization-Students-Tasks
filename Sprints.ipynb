```python
#creating a copy of the merged df to keep the original one safe
df_fe = merged_df.copy()
# Combining related columns into one variable
df_fe["TotalActivities"] = (
    df_fe["Extracurricular"] +
    df_fe["Sports"] +
    df_fe["Music"] +
    df_fe["Volunteering"]
)
print(df_fe["TotalActivities"].head())
#
```

Combining multiple activity variables into one to make it easier to visualize student involvement patterns.

```python
#removing the floats and creating three portions to work with using cut() to divide where q=percentiles
df_fe['Study_Habits'] = pd.qcut(df_fe['StudyTimeWeekly'], q=3, labels=['Low', 'Medium', 'High'])

print(df_fe['Study_Habits'])
```

Measuring academic performance per study hour.

Helps visualize how effectively students convert study time into performance.

It measures how effectively a student converts study time into performance
* Higher value = better efficiency (high GPA with less study time)
* Lower value = low efficiency (more study time but low GPA)

```python
df_fe["StudyEfficiency"] =np.divide(df_fe["GPA"], df_fe["StudyTimeWeekly"] + 1) # we are using +1 to avoid errors of dividing by zero which leads to infinity
print(df_fe["StudyEfficiency"])
```

Higher score indicates better attendance.

0 Absences results in a score of 1.0 which is perfect.

```python
df_fe["AttendanceScore"] = np.divide(1,df_fe["Absences"] + 1)
print(df_fe["AttendanceScore"].head())
```

**Milestone 4: Statistical Inference & Analytical Modeling**

STEP 1: Hypothesis testing

a) t-test

**Does tutoring affect GPA?**

* Null Hypothesis (H₀) - Tutoring has no effect on GPA

* Alternative Hypothesis (H₁) - Students with tutoring have different GPA


```python
from scipy.stats import ttest_ind
tutored = new_df[new_df["Tutoring"] == 1]["GPA"]
not_tutored = new_df[new_df["Tutoring"] == 0]["GPA"]
t_stat, p_value = ttest_ind(tutored, not_tutored)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

P-Value < 0.05 hence tutoring is one of the most important variables for a student's success.

Does gender affect GPA?

```python
male = new_df[new_df["Gender"] == 1]["GPA"]
female = new_df[new_df["Gender"] == 0]["GPA"]
t_stat, p_value = ttest_ind(male, female)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Gender does not affect GPA.

```python
high_support = new_df[new_df["ParentalSupport"] >= 2]["GPA"]
low_support = new_df[new_df["ParentalSupport"] < 2]["GPA"]
t_stat, p_value = ttest_ind(high_support, low_support)
print("T-statistic:", t_stat)
print("P-value:", p_value)
```

Students with high parental support (levels 2 and above) perform significantly better than those with low support.

b) . Chi-Square Test for Categorical relationship

Gender vs Tutoring

Hypotheses
* H₀ - No relationship between variables
* H₁ - Variables are related

```python
from scipy.stats import chi2_contingency
table = pd.crosstab(df["Gender"], df["Tutoring"])
chi2, p, dof, expected = chi2_contingency(table)
print("Chi-square:", chi2)
print("P-value:", p)
```

There is no significant link between Gender and Tutoring since p-value is greater than 0.005

**STEP 2: Correlation and regression analysis**

a) Correlation analysis


Relationship between StudyEfficiency and GPA

Does study efficiency relate to academic performance?

* H₀ - No relationship between StudyEfficiency and GPA
* H₁ - There is a relationship between StudyEfficiency and GPa




```python
from scipy.stats import pearsonr
corr, p_value = pearsonr(new_df["StudyEfficiency"], new_df["GPA"])
print("Correlation:", corr)
print("P-value:", p_value)
```

The correlation shows that there is a positive relationship but is weak to moderate strength

Hence as StudyEfficiency increases, GPA tends to increase slightly

b) Linear regression

```python
from sklearn.linear_model import LinearRegression
X = new_df[["StudyEfficiency"]]
y = new_df["GPA"]
model = LinearRegression()
model.fit(X, y)
print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0]) #slope
```

```python
plt.scatter(new_df["StudyEfficiency"], new_df["GPA"], alpha=0.5)

plt.plot(new_df["StudyEfficiency"],
         model.predict(X),
         color="red")

plt.title("Regression: Study Efficiency vs GPA")
plt.xlabel("Study Efficiency")
plt.ylabel("GPA")
plt.show()
```

For every 1 unit increase in StudyEfficiency, GPA increases by 0.9925 on average

Predicted GPA when StudyEfficiency = 0 is 1.637

A linear regression model was used to analyze the relationship between Study Efficiency and GPA. The model estimates how changes in study efficiency influence academic performance. The scatter plot shows actual data points while the regression line represents predicted values, allowing evaluation of the strength of the relationship.
